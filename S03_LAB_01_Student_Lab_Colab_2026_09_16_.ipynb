## 🧭 Worked contract and classification | 15 minutes
    In this guided example a usable row must have nonblank IDs, positive integer quantity,
    nonnegative decimal price/discount, and a discount no greater than quantity * price.
    A repeated line_id with different business values is a conflict requiring owner review.
    An exact repeated row can be retained once for a clean teaching result while both raw rows remain.
    **Safe casting:** validate text inside the same `CASE` expression that converts it. A `WHERE` filter alone
    does not promise that PostgreSQL will filter `unknown` before attempting `unknown::integer`.
    The following classification is a demonstration; transfer the approach to the different independent export.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
ck.execute("""
CREATE OR REPLACE VIEW gc3260_l01.classified AS
WITH checked AS (
 SELECT *, CASE
   WHEN nullif(line_id,'') IS NULL OR nullif(order_id,'') IS NULL OR nullif(product_id,'') IS NULL THEN false
   WHEN quantity !~ '^[0-9]+$' OR unit_price !~ '^[0-9]+(\.[0-9]+)?$'
     OR line_discount !~ '^[0-9]+(\.[0-9]+)?$' THEN false
   ELSE quantity::numeric > 0 AND line_discount::numeric <= quantity::numeric*unit_price::numeric
 END AS valid
 FROM gc3260_l01.raw_lines
), ranked AS (
 SELECT *, row_number() OVER(PARTITION BY line_id,order_id,product_id,quantity,unit_price,line_discount
                            ORDER BY _source_row) AS exact_copy_number FROM checked
)
SELECT *, CASE WHEN NOT valid THEN 'quarantine'
               WHEN exact_copy_number > 1 THEN 'exact_repeat' ELSE 'candidate' END AS disposition
FROM ranked
""")
display(ck.query("SELECT disposition, count(*) AS rows FROM gc3260_l01.classified GROUP BY 1 ORDER BY 1"))
````
