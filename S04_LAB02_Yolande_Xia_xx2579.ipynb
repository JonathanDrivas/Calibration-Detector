## 🎯 The decision and the core concepts

Turn four raw extracts into a model that answers Priya's question the same way every time.

- **Grain:** the exact meaning of one row. Every other modeling decision depends on it.
- **Fact:** a measurement at a declared grain, such as quantity or merchandise amount.
- **Dimension:** descriptive context for a fact, such as product category.
- **Business key:** an identifier that arrived from the source, such as `product_id`.
- **Surrogate key:** an identifier the warehouse generates, such as `product_key`.
- **CTE:** a named intermediate query. It has its own grain, which you should be able to state.
- **Window function:** a calculation across related rows that does not collapse them.
- **Referential integrity:** every foreign key resolves to the intended parent row.
- **Reconciliation:** comparing two independently defined quantities at the same scope.

## 🧠 SQL memory refresh | optional after the timed path

Three small patterns carry most of this lab. Each one connects syntax you may remember to a modeling risk.

| Pattern | Plain-language purpose | Risk it controls |
|---|---|---|
| **Precise cast** `::number(12,2)` | Turn source text into a number and keep cents. | A bare `::number` rounds cents away without any error. |
| **JOIN with a known cardinality** | Bring header fields to each line. | A header amount repeats once per line, so a sum inflates. |
| **Conditional aggregation** `SUM(IFF(...))`, `COUNT_IF(...)` | Put several filtered totals side by side at one grain. | Separate queries drift apart in their filters. |

### How to read a query before running it

1. **Result grain:** What does one returned row represent?
2. **Sources and joins:** Which table sets the grain, and can a join repeat its rows?
3. **Filter:** Which rows are allowed into the result?
4. **Shape:** Are rows kept individually, grouped, or compared through a window?
5. **Order and size:** How will the result stay readable?

`IFF(condition, value_if_true, value_if_false)` chooses a value row by row; `SUM` then adds the chosen values. The result keeps **one row per day** while showing web and store demand side by side. `COUNT_IF` counts the rows that satisfy a condition.

### Read the next query in two layers

1. The CTE named `daily` changes the grain from one order line to **one row per order date**, excluding cancelled orders.
2. The outer query keeps that daily grain and adds two window calculations:
   - a **running total**, which adds every row from the first date up to the current row; and
   - `LAG`, which looks one row back in date order so you can compute a change.

A window function does not collapse rows. The daily grain is the same before and after the window is applied.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
# 🎯 Purpose: answer the question stated immediately above.
# SQL uses one clause per line so the reading order is visible.
# Learn more: https://docs.snowflake.com/en/sql-reference/constructs
window_result = SQL("""
-- SQL family: DQL. CTE grain: one row per order date. Final grain: the same.
WITH daily AS (
  SELECT order_date,
         sum(merchandise_amount) AS amount
  FROM gc3260_l02.fct_order_lines
  WHERE status <> 'cancelled'
  GROUP BY order_date
)
SELECT order_date,
       amount,
       sum(amount) OVER (ORDER BY order_date
                         ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS cumulative_amount,
       amount - lag(amount) OVER (ORDER BY order_date)                     AS change_from_previous_observed_day
FROM daily
ORDER BY order_date
""")
# ✅ Expected: a labeled table. Read its row grain before interpreting values.
display(window_result)
````

### Rank categories without losing the totals

`DENSE_RANK()` numbers the groups by demand. Because the ranking runs **after** `GROUP BY`, its `ORDER BY` can use the aggregated amount. Read the grain: one row per product category.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
# 🎯 Purpose: answer the question stated immediately above.
# SQL uses one clause per line so the reading order is visible.
# Learn more: https://docs.snowflake.com/en/sql-reference/constructs
category_rank = SQL("""
-- SQL family: DQL. Result grain: one row per product category.
SELECT d.category,
       sum(f.merchandise_amount)                                      AS ordered_merchandise_amount,
       dense_rank() OVER (ORDER BY sum(f.merchandise_amount) DESC)   AS demand_rank
FROM gc3260_l02.fct_order_lines f
JOIN gc3260_l02.dim_product d USING (product_key)
WHERE f.status <> 'cancelled'
GROUP BY d.category
ORDER BY demand_rank
""")
# ✅ Expected: a labeled table. Read its row grain before interpreting values.
display(category_rank)
````

### Before you run the next cell: what it proves

The first query computes the demand amount twice, by two independent routes, at the same scope:

1. **Source route:** raw lines joined to raw orders, recomputing `quantity * unit_price - line_discount`.
2. **Model route:** the stored `merchandise_amount` in the fact table.

Equal totals are **necessary**. They are not **sufficient**: two errors can cancel inside a grand total. The second query therefore repeats the comparison for every order and counts the orders that disagree.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
# 🎯 Purpose: answer the question stated immediately above.
# SQL uses one clause per line so the reading order is visible.
# Learn more: https://docs.snowflake.com/en/sql-reference/constructs
reconciliation = SQL("""
-- SQL family: DQL. Result grain: one row for the whole guided case.
SELECT
  (SELECT sum(l.quantity::integer * l.unit_price::number(12,2) - l.line_discount::number(12,2))
   FROM gc3260_l02.raw_lines l
   JOIN gc3260_l02.raw_orders o USING (order_id)
   WHERE o.status <> 'cancelled')                              AS source_amount,
  (SELECT sum(merchandise_amount)
   FROM gc3260_l02.fct_order_lines
   WHERE status <> 'cancelled')                                AS fact_amount
""")
# ✅ Expected: a labeled table. Read its row grain before interpreting values.
display(reconciliation)
````

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
# 🎯 Purpose: answer the question stated immediately above.
# SQL uses one clause per line so the reading order is visible.
# Learn more: https://docs.snowflake.com/en/sql-reference/constructs
order_reconciliation = SQL("""
-- SQL family: DQL. Two CTEs at order grain, compared order by order.
WITH source_orders AS (
  SELECT l.order_id,
         sum(l.quantity::integer * l.unit_price::number(12,2) - l.line_discount::number(12,2)) AS source_amount
  FROM gc3260_l02.raw_lines l
  GROUP BY l.order_id
), fact_orders AS (
  SELECT order_id,
         sum(merchandise_amount) AS fact_amount
  FROM gc3260_l02.fct_order_lines
  GROUP BY order_id
)
SELECT count(*)                                                        AS orders_compared,
       count_if(s.order_id IS NULL OR f.order_id IS NULL)              AS orders_on_one_side_only,
       count_if(s.source_amount <> f.fact_amount)                      AS orders_with_variance
FROM source_orders s
FULL OUTER JOIN fact_orders f ON f.order_id = s.order_id
""")
# ✅ Expected: a labeled table. Read its row grain before interpreting values.
display(order_reconciliation)
````
