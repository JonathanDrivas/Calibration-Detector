## 🧠 SQL memory refresh | 6 minutes, low stakes

SQL is a language for telling a database **what result or change you need**. The database decides how to carry out that request. Before the contract work, reconnect the syntax you may remember to the job each statement performs.

| SQL family | Plain-language purpose | Example in this lab | Who performs it |
|---|---|---|---|
| **DQL: Data Query Language** | Read and summarize data without changing stored rows. | `SELECT`, `WHERE`, `GROUP BY`, `HAVING`, `ORDER BY`, `LIMIT` | You practice this below. |
| **DDL: Data Definition Language** | Create or change database structures. | `CREATE OR REPLACE VIEW` for the classification view | A supplied guided cell, then your assigned SQL work |

### How the classification SQL works

**SQL families in the next cell:** `CREATE OR REPLACE VIEW` is **DDL** because it defines a reusable database object. The `SELECT` inside the view is **DQL** because it specifies the rows and derived fields the view returns. The statement does not delete or overwrite the faithful raw table.

Read the next provided cell in three layers:

1. `checked` applies structural and business rules while protecting numeric casts.
2. `key_versions` detects a `line_id` attached to different business payloads.
3. `ranked` numbers rows that are exact matches across the complete business payload.
4. the final `CASE` assigns one mutually exclusive disposition and a reason.

Do not read the final counts as a grade by themselves. Read them as evidence that must later reconcile to the raw total and be explained to the manager.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
ck.execute("""
CREATE OR REPLACE VIEW gc3260_l01.classified AS
WITH checked AS (
 SELECT *, CASE
   WHEN nullif(line_id,'') IS NULL OR nullif(order_id,'') IS NULL OR nullif(product_id,'') IS NULL THEN false
   WHEN quantity !~ '^[0-9]+$' OR unit_price !~ '^[0-9]+(\\.[0-9]+)?$'
     OR line_discount !~ '^[0-9]+(\\.[0-9]+)?$' THEN false
   ELSE quantity::numeric > 0 AND line_discount::numeric <= quantity::numeric*unit_price::numeric
 END AS valid
 FROM gc3260_l01.raw_lines
), key_versions AS (
 SELECT line_id,
        count(DISTINCT (order_id,product_id,quantity,unit_price,line_discount)) AS line_versions
 FROM checked
 GROUP BY line_id
), ranked AS (
 SELECT c.*, k.line_versions,
        row_number() OVER(
          PARTITION BY c.line_id,c.order_id,c.product_id,c.quantity,c.unit_price,c.line_discount
          ORDER BY c._source_row
        ) AS exact_copy_number
 FROM checked c
 JOIN key_versions k USING (line_id)
)
SELECT *, CASE WHEN NOT valid THEN 'quarantine'
               WHEN line_versions > 1 THEN 'quarantine'
               WHEN exact_copy_number > 1 THEN 'exact_repeat'
               ELSE 'candidate' END AS disposition,
          CASE WHEN NOT valid THEN 'contract_rule_failure'
               WHEN line_versions > 1 THEN 'conflicting_line_id'
               WHEN exact_copy_number > 1 THEN 'exact_repeat'
               ELSE 'candidate' END AS disposition_reason
FROM ranked
""")
ck.execute("""
CREATE OR REPLACE VIEW gc3260_l01.candidate_lines AS
SELECT * FROM gc3260_l01.classified WHERE disposition = 'candidate'
""")
ck.execute("""
CREATE OR REPLACE VIEW gc3260_l01.quarantine_lines AS
SELECT * FROM gc3260_l01.classified WHERE disposition <> 'candidate'
""")
display(ck.query("""
SELECT disposition, disposition_reason, count(*) AS rows
FROM gc3260_l01.classified
GROUP BY 1, 2
ORDER BY 1, 2
"""))
````

### Exact repeat versus conflicting business key

These cases can look similar in a row count but require different decisions:

- **Exact repeat:** the same `line_id` and the same business payload appear more than once. This may be a resend. Retain all raw copies, but one clean representative may be eligible for downstream use.
- **Key conflict:** the same `line_id` appears with different order, product, quantity, price, or discount values. The system cannot know which version is authoritative, so all conflicting versions require quarantine and an owner decision.

`SELECT DISTINCT` is not a safe repair for a key conflict because it can hide the disagreement without resolving it.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
# 🎯 Purpose: answer the question stated immediately above.
# SQL uses one clause per line so the reading order is visible.
# Learn more: https://www.postgresql.org/docs/17/tutorial-sql.html
conflicts = SQL("""
SELECT line_id, count(DISTINCT (order_id,product_id,quantity,unit_price,line_discount)) AS versions
FROM gc3260_l01.raw_lines GROUP BY line_id
HAVING count(DISTINCT (order_id,product_id,quantity,unit_price,line_discount)) > 1
""")
# ✅ Expected: a labeled table. Read its row grain before interpreting values.
display(conflicts)
````

### Reconciliation is a conservation check

For the guided classification, every raw row must appear in exactly one disposition:

`raw rows = candidate rows + quarantine rows + exact-repeat rows`

The operational `quarantine_lines` view contains every noncandidate row, including exact repeats. The reconciliation keeps contract failures or conflicts separate from exact repeats so the business reason remains visible.

The reconciliation variance is the raw count minus the three mutually exclusive disposition counts. It must equal zero. Zero variance means **every row is accounted for**; it does not mean the batch is defect-free or ready for use. Read the defect counts and unresolved conflicts before recommending promotion.

````python
# PROVIDED: run unchanged. You are not graded on writing this Python.
# Read its comments to learn why it runs, what it changes, and what output to inspect.
display(ck.query("""
WITH counts AS (
  SELECT count(*) AS raw_rows,
         count(*) FILTER (WHERE disposition='candidate') AS candidate_rows,
         count(*) FILTER (WHERE disposition='quarantine') AS contract_quarantine_rows,
         count(*) FILTER (WHERE disposition='exact_repeat') AS exact_repeat_rows
  FROM gc3260_l01.classified
)
SELECT *,
       contract_quarantine_rows + exact_repeat_rows AS withheld_rows,
       raw_rows - candidate_rows - contract_quarantine_rows - exact_repeat_rows AS variance
FROM counts
"""))
ck.save_evidence("S03", "guided_raw_provenance", raw_evidence)
````

````python
# ✍️ STUDENT SQL: edit only the SQL inside the triple quotes.
# Task: Create candidate and quarantine views in gc3260_l01_ind from raw_lines. Preserve raw rows; handle invalid values, exact repeats and conflicting identities in SQL.
# Scaffold: most of the structure is supplied. Replace every __REPLACE_...__ token.
# Read each SQL comment. It explains the result grain, clause, and success condition.
# Learn more: https://www.postgresql.org/docs/17/tutorial-sql.html
display(SQL("""
-- SQL FAMILY: DDL creates or replaces views; the SELECT inside each view is DQL.
-- RESULT GRAIN: one classified row for every incoming order-line record.
-- PURPOSE: preserve the raw row, add safe numeric interpretations, and assign a disposition.
-- The student-defined rule and auditability explanation are completed below.

CREATE OR REPLACE VIEW gc3260_l01_ind.classified_lines AS
-- typed safely converts text only when its pattern is valid.
WITH typed AS (
  SELECT *,
    CASE WHEN quantity ~ '^[0-9]+$' THEN quantity::integer END AS quantity_number,
    CASE WHEN unit_price ~ '^[0-9]+(\\.[0-9]+)?$' THEN unit_price::numeric END AS unit_price_number,
    CASE WHEN line_discount ~ '^[0-9]+(\\.[0-9]+)?$' THEN line_discount::numeric END AS discount_number
  FROM gc3260_l01_ind.raw_lines
),
-- conflicts counts distinct business payloads for each proposed line_id.
conflicts AS (
  SELECT line_id, count(DISTINCT (order_id, product_id, quantity, unit_price, line_discount)) AS line_versions
  FROM typed
  GROUP BY line_id
),
-- marked numbers byte-equivalent repeats without deleting any source row.
marked AS (
  SELECT typed.*, conflicts.line_versions,
    row_number() OVER (PARTITION BY line_id, order_id, product_id, quantity, unit_price, line_discount
                       ORDER BY _source_row) AS exact_copy_number
  FROM typed
  LEFT JOIN conflicts USING (line_id)
)
SELECT *, CASE
  WHEN nullif(line_id, '') IS NULL OR nullif(order_id, '') IS NULL OR nullif(product_id, '') IS NULL
    THEN 'missing_identifier'
  WHEN quantity_number IS NULL OR quantity_number <= 0
    OR unit_price_number IS NULL OR unit_price_number < 0
    OR discount_number IS NULL OR discount_number < 0
    THEN 'invalid_numeric_value'
  WHEN discount_number > quantity_number * unit_price_number THEN 'discount_exceeds_value'
  WHEN line_versions > 1 THEN 'conflicting_line_id'
  WHEN exact_copy_number > 1 THEN 'exact_repeat'
  -- STUDENT STEP 1: replace the token with one Boolean condition for a manager-relevant source rule.
  -- Example shape only: WHEN <condition> THEN '<short_reason>'
  WHEN line_id !~ '^L[0-9]+$'
  OR order_id !~ '^O[0-9]+$'
  OR product_id !~ '^P[0-9]+$'
THEN 'manager_rule_failure'
  ELSE 'candidate'
END AS disposition
FROM marked;

CREATE OR REPLACE VIEW gc3260_l01_ind.candidate_lines AS
SELECT * FROM gc3260_l01_ind.classified_lines WHERE disposition = 'candidate';

CREATE OR REPLACE VIEW gc3260_l01_ind.quarantine_lines AS
SELECT * FROM gc3260_l01_ind.classified_lines WHERE disposition <> 'candidate';
-- STUDENT STEP 2: replace the token with your auditability explanation.
-- Why raw quarantine is retained: Original values and provenance remain available for investigation, correction, and reproducible reclassification without overwriting the source.

""", save_as="work/S03/sql/01_model.sql"))
````
